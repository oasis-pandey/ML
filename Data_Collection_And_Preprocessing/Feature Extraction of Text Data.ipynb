Feature Extraction

The mapping from textual data to real valued vectors is called feature extraction.
(basically converting the text data to numerical data to send to the ML model)

BOW: Bag Of Words. --> List of unique words in the text corpus

TF-IDF: Term Frequency-Inverse Document Frequency --> To count the number of times each word appears in a document.

Term Frequency = (Number of times a term t appears in a document)/(Total number of terms in the document)

Inverse Document Frequency(IDF) = log(N/n), where, N is the number of documents in a collection and n is the number of documents a term t has appeared in.

The IDF value of a rare word is high and the IDF of a frequent word is low.

TF-IDF value of a term = TF * IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Sample text corpus
documents = [
    "The cat sat on the mat.",
    "The cat chased the rat.",
    "The dog sat on the mat."
]

# Initialize the TF-IDF Vectorizer
vectorizer = TfidfVectorizer()

# Transform the documents into TF-IDF features
tfidf_matrix = vectorizer.fit_transform(documents)

# Get the feature names (terms)
terms = vectorizer.get_feature_names_out()

# Convert the TF-IDF matrix into a dense format for easy viewing
dense_matrix = tfidf_matrix.todense()

# Display the TF-IDF values for each term in each document
import pandas as pd
df = pd.DataFrame(dense_matrix, columns=terms)

print(df)
```
