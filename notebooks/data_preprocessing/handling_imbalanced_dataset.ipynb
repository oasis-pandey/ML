This is Highly imbalanced dataset.

0 -- > Legit Transactions

1 --> Fraud Transactions

```python
#seperating the legit and fraud transcations
legit = df[df['Class'] == 0]
fraud = df[df['Class']== 1]
```

```python
print(legit.shape)
```

```python
print(fraud.shape)
```

Under-Sampling:

we are building a sample dataset containing similar distribution of legit transactions and fraud transactions. So we have about 423 fraud and 233319 legit which is very imbalanced. we take similar amount of legit transcations, say for example 500

```python
#take 500 random sample from legit data to balance
legit_sample = legit.sample(n=500)
```

```python
legit_sample.shape
```

```python
#concatenate the legit_sample and fraud sample
new_df = pd.concat([legit_sample,fraud], axis=0)
```

```python
new_df.shape
```
