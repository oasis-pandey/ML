```python
from io import StringIO

csv_data = StringIO("""
age,income,city,spend
25,50000,NY,200
30,60000,LA,NaN
22,,NY,120
35,80000,Chicago,300
28,55000,,150
""")
```

```python
df = pd.read_csv(csv_data)
```

```python
df.dropna(subset = ["income"])
```

```python
df = pd.DataFrame({
    "age": [20, 22, None, 30, 28],
    "salary": [30000, None, 45000, 60000, None],
    "department": ["IT", "HR", "IT", None, "HR"]
})
df
```

```python
df.isna().sum()
```

```python
df["age"] = df["age"].fillna(df["age"].mean())
```

```python
df["salary"] = df["salary"].fillna(df["salary"].median())
```

```python
df["department"] = df["department"].fillna(df["department"].mode()[0])
```

```python
df.isna().sum()
```

```python
df
```

```python
df.groupby("department")["salary"].mean()
```
