```python
expense = np.array([100, 50, 120, 200, 250, 20])
total_expense = np.sum(expense)
average_expense = np.mean(expense)
highest_expense = np.max(expense)
taxed_expense = expense * 1.1
print(f"Total:{total_expense}, Average: {average_expense}, Highest: {highest_expense}, Taxed: {taxed_expense}")
```

Let's Look at 2D arrays now. 

```python
kharcha = np.array([
    [100, 50, 20],   # Day 1: food, transport, misc
    [120, 60, 30],   # Day 2
    [90,  40, 25]    # Day 3
])
print(kharcha)
```

```python
print(kharcha.shape)
```

```python
print(kharcha[0])
```

```python
print(kharcha[1,:])
```

```python
print(np.sum(kharcha, axis=1))
```

```python
scaled = kharcha / np.max(kharcha)
print(scaled)
```

```python
        #Categories #A    #B   #C
weekly = np.array([[100, 200, 300], #Monday
                   [120, 220, 290], #Tuesday
                   [110,240, 200]]) #Wednesday
```

```python
total_A = np.sum(weekly[:,0])
print(total_A)
```

```python
total_B = np.sum(weekly[:,1])
print(total_B)
```

```python
total_C = np.sum(weekly[:,2])
print(total_C)
```

```python
average_monday = np.mean(weekly[0,:])
print(average_monday)
```

```python
average_Tuesday = np.mean(weekly[1,:])
print(average_Tuesday)
```

```python
average_Wednesday = np.mean(weekly[2,:])
print(average_Wednesday)
```

```python
highest_weekly_expense = np.max(weekly)
print(highest_weekly_expense)
```

At this point we have completed learning Numpy for 1D and 2D arrays with basic commands. Now, lets move on to the last topic which is Reshaping and broadcasting. Then, we can mark this as completed and move towards learning Pandas. 

```python
data = np.array([100,200,300,400,500,600])
print(data.shape)
```

```python
reshaped = data.reshape(3,2)
print(reshaped)
print(reshaped.shape)
```

```python
auto = data.reshape(3,-1)
print(auto)
```

```python
weekly = np.array([
    [100, 200, 300],
    [120, 220, 290],
    [110, 240, 200]
])

tax = np.array([1.1, 1.05, 1.2])

total_with_tax = weekly * tax
print(total_with_tax)
```

```python
discount = np.array([10, 20, 30])
print(weekly - discount)
```

```python
daily_fee = np.array([[10], [10], [10]])
print(weekly + daily_fee)
```
